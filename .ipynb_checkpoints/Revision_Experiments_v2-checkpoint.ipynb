## Part A. Evaluation of the frozen models (no training)
### Cell 1. Reload environment, baselines and frozen settings (copied unchanged from the frozen notebook)

```python
import numpy as np
import gymnasium as gym
from gymnasium import spaces
from dataclasses import dataclass
from enum import IntEnum
from stable_baselines3 import PPO
from scipy import stats

FIBONACCI_POINTS = [1, 2, 3, 5, 8, 13]
MAX_EFFORT = max(FIBONACCI_POINTS)
ITEM_NAME_POOL = ["Bulk CSV export", "SSO login (SAML)", "Dark mode UI", "Async job queue",
    "Slack integration", "Custom dashboard widgets", "Audit log export", "Two-factor authentication",
    "Webhook retries", "API rate limiting", "Kanban swimlanes", "Time-tracking report",
    "Mobile push notifications", "Bulk item archive", "Custom fields for tickets", "SLA breach alerts",
    "Guest viewer role", "Recurring task templates", "Dependency graph view", "CSV import validation",
    "Public roadmap page", "Sprint burndown chart", "Email digest summaries", "Attachment virus scanning",
    "Search autocomplete", "Bulk reassign owner", "Custom status workflows", "Team capacity planner",
    "Item history timeline", "Keyboard shortcuts", "Offline mode (PWA)", "Multi-currency billing",
    "Usage analytics dashboard", "SSO for SCIM provisioning", "Comment @mentions", "Item templates library",
    "Read-only API tokens", "Data residency (EU region)", "Custom SLA policies", "Bulk label editor"]

@dataclass
class BacklogItem:
    item_id: int; name: str; business_value: float; effort_raw: int; effort: float
    risk: float; dependency_count: int; demand_signal: float; urgency: float; age: int = 0
    def to_feature_vector(self):
        return np.array([self.business_value, self.effort, self.risk, min(self.dependency_count, 3)/3.0,
                          self.demand_signal, self.urgency, min(self.age, 20)/20.0], dtype=np.float32)

class BacklogGenerator:
    def __init__(self, seed=None):
        self.rng = np.random.default_rng(seed); self._next_id = 0
        self._name_cycle = list(ITEM_NAME_POOL); self.rng.shuffle(self._name_cycle); self._name_idx = 0
    def _next_name(self):
        name = self._name_cycle[self._name_idx % len(self._name_cycle)]; self._name_idx += 1; return name
    def generate_item(self):
        item_id = self._next_id; self._next_id += 1
        business_value = float(self.rng.beta(2, 2)); effort_raw = int(self.rng.choice(FIBONACCI_POINTS))
        effort = effort_raw / MAX_EFFORT; risk = float(self.rng.beta(1.5, 3))
        dependency_count = int(min(self.rng.poisson(0.6), 3)); demand_signal = float(self.rng.beta(2, 3))
        urgency = float(self.rng.beta(1.5, 4))
        return BacklogItem(item_id=item_id, name=self._next_name(), business_value=round(business_value,3),
            effort_raw=effort_raw, effort=round(effort,3), risk=round(risk,3),
            dependency_count=dependency_count, demand_signal=round(demand_signal,3),
            urgency=round(urgency,3), age=0)
    def generate_backlog(self, n_items):
        return [self.generate_item() for _ in range(n_items)]

MAX_ITEMS = 40; N_FEATURES = 7; EPISODE_LENGTH = 12; SHOCK_PROBABILITY = 0.20
W1_VALUE = 1.0; W2_DEPENDENCY = 0.5; W3_CAPACITY = 1.0; W4_URGENCY = 0.3; URGENCY_THRESHOLD = 0.7

class BacklogPrioritisationEnv(gym.Env):
    metadata = {"render_modes": ["human"]}
    def __init__(self, n_items=30, capacity_points=20, seed=None):
        super().__init__()
        self.n_items_target = n_items; self.capacity_points = capacity_points
        self.generator = BacklogGenerator(seed=seed)
        obs_dim = MAX_ITEMS * N_FEATURES + 3
        self.observation_space = spaces.Box(low=0.0, high=1.0, shape=(obs_dim,), dtype=np.float32)
        self.action_space = spaces.MultiDiscrete([3] * MAX_ITEMS)
        self.items = []; self.cycle = 0; self.last_shock = 0
    def reset(self, *, seed=None, options=None):
        super().reset(seed=seed)
        if seed is not None: self.generator = BacklogGenerator(seed=seed)
        self.items = self.generator.generate_backlog(self.n_items_target)
        self.cycle = 0; self.last_shock = 0
        return self._get_obs(), self._get_info()
    def step(self, action):
        shock_applied = self._maybe_apply_shock()
        reward, info = self._compute_reward_and_advance(action)
        self.cycle += 1; terminated = False; truncated = self.cycle >= EPISODE_LENGTH
        info["shock_applied"] = shock_applied; info["cycle"] = self.cycle
        return self._get_obs(), reward, terminated, truncated, info
    def _maybe_apply_shock(self):
        if self.generator.rng.random() >= SHOCK_PROBABILITY or not self.items:
            self.last_shock = 0; return None
        self.last_shock = 1
        shock_type = self.generator.rng.choice(["demand_spike", "dependency_resolved", "capacity_change"])
        target = self.items[self.generator.rng.integers(0, len(self.items))]
        if shock_type == "demand_spike": target.demand_signal = min(1.0, target.demand_signal + 0.4)
        elif shock_type == "dependency_resolved": target.dependency_count = max(0, target.dependency_count - 1)
        elif shock_type == "capacity_change":
            delta = self.generator.rng.choice([-5, 5]); self.capacity_points = max(5, self.capacity_points + delta)
        return shock_type
    def _compute_reward_and_advance(self, action):
        value_delivered = 0.0; dependency_violation = 0; urgency_delay = 0.0; now_effort = 0
        now_items, next_items, later_items = [], [], []
        for idx, item in enumerate(self.items):
            tier = int(action[idx]) if idx < len(action) else 0
            if tier == 2:
                now_items.append(item); value_delivered += item.business_value; now_effort += item.effort_raw
                if item.dependency_count > 0: dependency_violation += 1
            elif tier == 1: next_items.append(item)
            else: later_items.append(item)
        capacity_overrun = max(0, now_effort - self.capacity_points)
        for item in self.items:
            if item.urgency > URGENCY_THRESHOLD and item not in now_items: urgency_delay += item.urgency
        reward = (W1_VALUE*value_delivered - W2_DEPENDENCY*dependency_violation
                  - W3_CAPACITY*capacity_overrun - W4_URGENCY*urgency_delay)
        completed_ids = {it.item_id for it in now_items}
        self.items = [it for it in self.items if it.item_id not in completed_ids]
        for it in self.items: it.age += 1
        n_new = max(0, self.n_items_target - len(self.items))
        self.items.extend(self.generator.generate_backlog(n_new))
        info = {"value_delivered": value_delivered, "dependency_violation": dependency_violation,
                "capacity_overrun": capacity_overrun, "urgency_delay": urgency_delay,
                "n_now": len(now_items), "n_next": len(next_items), "n_later": len(later_items)}
        return reward, info
    def _get_obs(self):
        mat = np.zeros((MAX_ITEMS, N_FEATURES), dtype=np.float32)
        for i, item in enumerate(self.items[:MAX_ITEMS]): mat[i] = item.to_feature_vector()
        global_feats = np.array([min(self.capacity_points, 50)/50.0, min(self.cycle, EPISODE_LENGTH)/EPISODE_LENGTH,
                                  float(self.last_shock)], dtype=np.float32)
        return np.concatenate([mat.flatten(), global_feats])
    def _get_info(self):
        return {"n_active_items": len(self.items), "capacity_points": self.capacity_points}

class Tier(IntEnum):
    LATER = 0; NEXT = 1; NOW = 2

def rice_score(item, confidence_risk_threshold=0.7):
    reach = item.demand_signal; impact = item.business_value
    confidence = 0.5 if item.risk > confidence_risk_threshold else 0.8
    effort = max(item.effort, 1e-6)
    return (reach * impact * confidence) / effort

def wsjf_score(item):
    cost_of_delay = item.business_value + item.urgency + (1.0 - item.risk)
    duration = max(item.effort, 1e-6)
    return cost_of_delay / duration

def assign_tiers(items, score_fn, capacity_points, next_band_fraction=0.5):
    scored = sorted(((it, score_fn(it)) for it in items), key=lambda x: x[1], reverse=True)
    tiers = {}; used_now = 0; used_next = 0; next_budget = capacity_points * next_band_fraction
    for item, score in scored:
        if used_now + item.effort_raw <= capacity_points:
            tiers[item.item_id] = Tier.NOW; used_now += item.effort_raw
        elif used_next + item.effort_raw <= next_budget:
            tiers[item.item_id] = Tier.NEXT; used_next += item.effort_raw
        else: tiers[item.item_id] = Tier.LATER
    return tiers, scored

def policy_action_from_tiers(env, tiers_dict):
    action = np.zeros(MAX_ITEMS, dtype=int)
    for idx, item in enumerate(env.items): action[idx] = int(tiers_dict.get(item.item_id, 0))
    return action

TRAIN_SEEDS = [10, 11, 12, 13, 14, 15, 16, 17]
NEW_TEST_SEEDS = list(range(200, 215))
N_CYCLES = 12; CAPACITY = 20; MODEL_DIR = "models"

def run_episode_static(policy_name, seed):
    env = BacklogPrioritisationEnv(n_items=30, capacity_points=CAPACITY, seed=seed)
    obs, info = env.reset(seed=seed)
    rewards = []
    for cycle in range(N_CYCLES):
        if policy_name == "random": action = env.action_space.sample()
        else:
            score_fn = rice_score if policy_name == "rice" else wsjf_score
            tiers, _ = assign_tiers(env.items, score_fn, env.capacity_points)
            action = policy_action_from_tiers(env, tiers)
        obs, reward, term, trunc, info = env.step(action)
        rewards.append(reward)
        if term or trunc: break
    return rewards

baseline_results = {"random": [], "rice": [], "wsjf": []}
for seed in NEW_TEST_SEEDS:
    for policy in baseline_results:
        baseline_results[policy].append(sum(run_episode_static(policy, seed)))

print("Reloaded successfully.")
print(f"RICE: {np.mean(baseline_results['rice']):.2f}  WSJF: {np.mean(baseline_results['wsjf']):.2f}")

import os
missing = [s for s in TRAIN_SEEDS if not os.path.exists(f"{MODEL_DIR}/ppo_seed{s}.zip")]
if missing:
    print(f"WARNING: missing model files for seeds {missing}")
else:
    print(f"All 8 trained models found on disk. Ready for Cell 7 evaluation.")
```

### Cell 2. Shared helpers

```python
import json, time, os
import torch
from scipy import stats
from stable_baselines3 import PPO

RESULTS_A = {}
if os.path.exists("revision_partA.json"):          # keep results from earlier runs if the kernel restarted
    RESULTS_A = json.load(open("revision_partA.json"))
def save_A():
    with open("revision_partA.json", "w") as f: json.dump(RESULTS_A, f, indent=2, default=float)
    print("(progress saved to revision_partA.json)")

def set_shock_probability(p):
    """The environment reads the global SHOCK_PROBABILITY at run time."""
    globals()["SHOCK_PROBABILITY"] = p

def set_capacity_weight(w):
    """The environment reads the global W3_CAPACITY at run time."""
    globals()["W3_CAPACITY"] = w

def get_now_probabilities(model, obs):
    obs_tensor = torch.as_tensor(obs).float().unsqueeze(0)
    with torch.no_grad():
        dist = model.policy.get_distribution(obs_tensor)
        return np.array([d.probs[0, 2].item() for d in dist.distribution])

def fill_by_score(env, scores, next_threshold=None):
    """Rank active items by score (high first) and put them in Now until capacity is full.
    Items that do not fit go to Next if score > next_threshold, otherwise Later.
    (Next and Later earn the same reward in this environment.)"""
    n = len(env.items)
    action = np.zeros(MAX_ITEMS, dtype=int)
    used = 0
    for idx in np.argsort(-np.asarray(scores[:n]), kind="stable"):
        item = env.items[idx]
        if used + item.effort_raw <= env.capacity_points:
            action[idx] = 2; used += item.effort_raw
        elif next_threshold is not None and scores[idx] > next_threshold:
            action[idx] = 1
    return action

def item_gain(item):
    """Immediate reward change from putting this one item in Now (ignoring capacity)."""
    g = W1_VALUE * item.business_value
    if item.dependency_count > 0: g -= W2_DEPENDENCY
    if item.urgency > URGENCY_THRESHOLD: g += W4_URGENCY * item.urgency  # avoids the urgency penalty
    return g

def oracle_action(env):
    """Per-sprint (myopic) knapsack optimum: best immediate reward for THIS sprint without overrun.
    Exact 0/1 knapsack by dynamic programming. Not a global optimum: it ignores future sprints."""
    items = env.items; cap = int(env.capacity_points)
    gains = [item_gain(it) for it in items]
    best = np.zeros(cap + 1); keep = np.zeros((len(items), cap + 1), dtype=bool)
    for i, it in enumerate(items):
        w = it.effort_raw
        if gains[i] <= 0: continue
        for c in range(cap, w - 1, -1):
            if best[c - w] + gains[i] > best[c]:
                best[c] = best[c - w] + gains[i]; keep[i, c] = True
    action = np.zeros(MAX_ITEMS, dtype=int); c = cap
    for i in range(len(items) - 1, -1, -1):
        if keep[i, c]:
            action[i] = 2; c -= items[i].effort_raw
    return action

def run_episode(decide, seed, env_cls=None):
    """decide(env, obs) -> action. Returns total reward and per-sprint diagnostics."""
    env_cls = env_cls or BacklogPrioritisationEnv
    env = env_cls(n_items=30, capacity_points=CAPACITY, seed=seed)
    obs, info = env.reset(seed=seed)
    total = 0.0; n_now = []; overrun = []; shock_flags = []; step_rewards = []
    for cycle in range(N_CYCLES):
        action = decide(env, obs)
        obs, reward, term, trunc, info = env.step(action)
        total += reward; step_rewards.append(reward)
        n_now.append(info["n_now"]); overrun.append(info["capacity_overrun"])
        shock_flags.append(info["shock_applied"] is not None)
        if term or trunc: break
    return dict(total=total, n_now=float(np.mean(n_now)), overrun=float(np.mean(overrun)),
                step_rewards=step_rewards, shocks=shock_flags)

def mean_over_scenarios(decide, env_cls=None, seeds=None):
    eps = [run_episode(decide, s, env_cls) for s in (seeds or NEW_TEST_SEEDS)]
    return dict(mean=float(np.mean([e["total"] for e in eps])),
                n_now=float(np.mean([e["n_now"] for e in eps])),
                overrun=float(np.mean([e["overrun"] for e in eps])), episodes=eps)

# Decision rules
def ppo_deterministic(model): return lambda env, obs: model.predict(obs, deterministic=True)[0]
def ppo_sampled(model):       return lambda env, obs: model.predict(obs, deterministic=False)[0]
def ppo_capacity_aware(model):
    return lambda env, obs: fill_by_score(env, get_now_probabilities(model, obs), next_threshold=0.3)

def holm(pvals):
    """Holm step-down adjustment with monotonicity (handles ties correctly)."""
    p = np.asarray(pvals, float); order = np.argsort(p); m = len(p)
    adj = np.empty(m); running = 0.0
    for rank, i in enumerate(order):
        running = max(running, min(1.0, (m - rank) * p[i])); adj[i] = running
    return adj.tolist()

boot_rng = np.random.default_rng(42)
def bootstrap_median_diff(seed_means, reference, n_boot=10000):
    x = np.asarray(seed_means)
    d = [np.median(boot_rng.choice(x, size=len(x), replace=True)) - reference for _ in range(n_boot)]
    return [float(np.percentile(d, 2.5)), float(np.percentile(d, 97.5))]

def compare_to_baselines(name, seed_means, margin=5.0):
    out = {"seed_means": [float(v) for v in seed_means], "mean": float(np.mean(seed_means)),
           "median": float(np.median(seed_means))}
    raw = []
    for bname, ref in [("RICE", RICE_MEAN), ("WSJF", WSJF_MEAN)]:
        diffs = np.asarray(seed_means) - ref
        w, p = stats.wilcoxon(diffs) if np.any(diffs != 0) else (np.nan, 1.0)
        ci = bootstrap_median_diff(seed_means, ref)
        out[bname] = {"ref_mean": ref, "W": float(w), "raw_p": float(p), "ci95_median_diff": ci,
                      "inside_margin": bool(ci[0] >= -margin and ci[1] <= margin)}
        raw.append(p)
    adj = holm(raw); out["RICE"]["holm_p"] = adj[0]; out["WSJF"]["holm_p"] = adj[1]
    print(f"\n{name}: mean {out['mean']:.2f}, median {out['median']:.2f}")
    for b in ["RICE", "WSJF"]:
        r = out[b]
        print(f"  vs {b} (mean {r['ref_mean']:.2f}): raw p={r['raw_p']:.4f}, Holm p={r['holm_p']:.4f}, "
              f"95% CI of median diff [{r['ci95_median_diff'][0]:.2f}, {r['ci95_median_diff'][1]:.2f}], "
              f"inside +/-{margin}: {r['inside_margin']}")
    return out

set_shock_probability(0.20); set_capacity_weight(1.0)
ppo_models = {s: PPO.load(f"{MODEL_DIR}/ppo_seed{s}", device="cpu") for s in TRAIN_SEEDS}
RICE_MEAN = float(np.mean(baseline_results["rice"])); WSJF_MEAN = float(np.mean(baseline_results["wsjf"]))
if "sanity" in RESULTS_A:                           # lets you continue from a later cell after a restart
    det = RESULTS_A["sanity"]["det_seed_means"]; cap = RESULTS_A["sanity"]["cap_seed_means"]
print(f"Loaded {len(ppo_models)} frozen models. RICE mean = {RICE_MEAN:.2f}, WSJF mean = {WSJF_MEAN:.2f}")
```

### Cell 4. Corrected statistics (C5, M5)
Compares against baseline **means** (not medians), applies Holm correctly, and adds Spearman and leave-one-seed-out correlations.

```python
RESULTS_A["stats"] = {"rice_mean": RICE_MEAN, "wsjf_mean": WSJF_MEAN,
                      "rice_minus_wsjf": RICE_MEAN - WSJF_MEAN,
                      "deterministic": compare_to_baselines("Deterministic decoding", det),
                      "capacity_aware": compare_to_baselines("Capacity-aware decoding", cap)}

# Training reward vs held-out reward
final_train = np.array([np.load(f"{MODEL_DIR}/convergence_seed{s}.npy")[-1][1] for s in TRAIN_SEEDS])
stoch = [mean_over_scenarios(ppo_sampled(ppo_models[s]))["mean"] for s in TRAIN_SEEDS]
corr = {}
for label, test_vals in [("deterministic", det), ("sampled", stoch)]:
    r, p = stats.pearsonr(final_train, test_vals); rho, prho = stats.spearmanr(final_train, test_vals)
    loo = [stats.pearsonr(np.delete(final_train, i), np.delete(test_vals, i))[0] for i in range(len(TRAIN_SEEDS))]
    i15 = TRAIN_SEEDS.index(15)
    rho_wo15 = stats.spearmanr(np.delete(final_train, i15), np.delete(test_vals, i15))[0]
    corr[label] = dict(pearson_r=float(r), pearson_p=float(p), spearman_rho=float(rho), spearman_p=float(prho),
                       loo_pearson_min=float(min(loo)), loo_pearson_max=float(max(loo)),
                       pearson_without_seed15=float(loo[i15]), spearman_without_seed15=float(rho_wo15))
    print(f"\nTraining reward vs {label} held-out reward: Pearson r={r:.3f} (p={p:.4f}), "
          f"Spearman rho={rho:.3f} (p={prho:.4f}); leave-one-out Pearson range [{min(loo):.3f}, {max(loo):.3f}]; "
          f"without seed 15: Pearson {loo[i15]:.3f}, Spearman {rho_wo15:.3f}")
RESULTS_A["correlation"] = corr
RESULTS_A["final_training_reward"] = final_train.tolist()
RESULTS_A["sampled_seed_means"] = stoch
print("\nSampled (stochastic) decoding seed means:", [round(v, 2) for v in stoch], f"mean {np.mean(stoch):.2f}")
save_A()
```

### Cell 5. What each decoding rule actually does (C3, M5)
Average number of items put in Now per sprint, average capacity overrun (story points above capacity), and average P(Now).

```python
diag = {}
for mode, rule in [("deterministic", ppo_deterministic), ("sampled", ppo_sampled), ("capacity_aware", ppo_capacity_aware)]:
    rows = [mean_over_scenarios(rule(ppo_models[s])) for s in TRAIN_SEEDS]
    diag[mode] = dict(mean_reward=float(np.mean([r["mean"] for r in rows])),
                      n_now_per_sprint=float(np.mean([r["n_now"] for r in rows])),
                      overrun_per_sprint=float(np.mean([r["overrun"] for r in rows])),
                      per_seed_n_now=[r["n_now"] for r in rows])
for name in ["rice", "wsjf"]:
    score_fn = rice_score if name == "rice" else wsjf_score
    r = mean_over_scenarios(lambda env, obs, f=score_fn: policy_action_from_tiers(env, assign_tiers(env.items, f, env.capacity_points)[0]))
    diag[name] = dict(mean_reward=r["mean"], n_now_per_sprint=r["n_now"], overrun_per_sprint=r["overrun"])

# Mean P(Now) on the first sprint of each test scenario, per seed
pnow = {}
for s in TRAIN_SEEDS:
    vals = []
    for ts in NEW_TEST_SEEDS:
        env = BacklogPrioritisationEnv(n_items=30, capacity_points=CAPACITY, seed=ts); obs, _ = env.reset(seed=ts)
        vals.append(get_now_probabilities(ppo_models[s], obs)[:len(env.items)].mean())
    pnow[s] = float(np.mean(vals))
diag["mean_p_now_by_seed"] = pnow
r_p, p_p = stats.spearmanr(list(pnow.values()), det)
diag["spearman_pnow_vs_deterministic"] = [float(r_p), float(p_p)]

print(f"{'Rule':<16}{'Reward':>10}{'Now items/sprint':>18}{'Overrun pts/sprint':>20}")
for k in ["deterministic", "sampled", "capacity_aware", "rice", "wsjf"]:
    d = diag[k]; print(f"{k:<16}{d['mean_reward']:>10.2f}{d['n_now_per_sprint']:>18.2f}{d['overrun_per_sprint']:>20.2f}")
print("\nMean P(Now) per seed:", {k: round(v, 3) for k, v in pnow.items()})
print(f"Spearman: mean P(Now) vs deterministic score: rho={r_p:.3f}, p={p_p:.4f}")
RESULTS_A["decoding_diagnostics"] = diag
save_A()
```

### Cell 6. Controls for capacity-aware decoding (C4)
Every rule below uses the **same capacity fill**. Only the ranking changes. If random ranking scores about the same as the trained network, the network is not adding much.

```python
controls = {}
def seeded_random_rule(rng):
    return lambda env, obs: fill_by_score(env, rng.random(len(env.items)))

# Seed-dependent rules: one value per training seed (8 values)
controls["ppo_trained_ranking"] = cap
controls["random_ranking"] = [mean_over_scenarios(seeded_random_rule(np.random.default_rng(1000 + s)))["mean"] for s in TRAIN_SEEDS]
untrained = []
for s in TRAIN_SEEDS:
    m = PPO("MlpPolicy", BacklogPrioritisationEnv(n_items=30, capacity_points=CAPACITY, seed=s), seed=s, device="cpu",
            policy_kwargs=dict(net_arch=dict(pi=[128, 128], vf=[128, 128])))
    untrained.append(mean_over_scenarios(ppo_capacity_aware(m))["mean"])
controls["untrained_network_ranking"] = untrained

# Fixed rules: one value (mean over the 15 scenarios)
fixed = {
    "value_only": lambda env, obs: fill_by_score(env, [it.business_value for it in env.items]),
    "value_per_effort": lambda env, obs: fill_by_score(env, [it.business_value / it.effort_raw for it in env.items]),
    "smallest_effort_first": lambda env, obs: fill_by_score(env, [-it.effort_raw for it in env.items]),
    "gain_per_effort_greedy": lambda env, obs: fill_by_score(env, [item_gain(it) / it.effort_raw for it in env.items]),
    "myopic_knapsack_oracle": lambda env, obs: oracle_action(env),
}
for k, rule in fixed.items(): controls[k] = mean_over_scenarios(rule)["mean"]
controls["rice"] = RICE_MEAN; controls["wsjf"] = WSJF_MEAN

print(f"{'Ranking (all with the same capacity fill)':<42}{'Mean':>9}{'Min seed':>10}{'Max seed':>10}")
for k, v in controls.items():
    if isinstance(v, list): print(f"{k:<42}{np.mean(v):>9.2f}{min(v):>10.2f}{max(v):>10.2f}")
    else: print(f"{k:<42}{v:>9.2f}")
w, p = stats.wilcoxon(np.array(cap) - np.array(controls["random_ranking"]))
print(f"\nTrained vs random ranking (paired by seed): Wilcoxon p={p:.4f}")
w2, p2 = stats.wilcoxon(np.array(cap) - np.array(untrained))
print(f"Trained vs untrained network (paired by seed): Wilcoxon p={p2:.4f}")
controls["trained_vs_random_p"] = float(p); controls["trained_vs_untrained_p"] = float(p2)
RESULTS_A["controls"] = controls
save_A()
```

### Cell 7. Do the methods handle surprises? Shocks on vs off (RQ1, M3)
Same 15 test scenarios, run once with shocks (probability 0.20) and once without (0.0). Also compares reward in sprints where a shock happened with sprints where none did.
Note: switching shocks off also changes which new items are generated later in the episode, because shocks use the same random generator.

```python
rules = {
    "ppo_capacity_aware": None,  # averaged over the 8 seeds below
    "rice": lambda env, obs: policy_action_from_tiers(env, assign_tiers(env.items, rice_score, env.capacity_points)[0]),
    "wsjf": lambda env, obs: policy_action_from_tiers(env, assign_tiers(env.items, wsjf_score, env.capacity_points)[0]),
    "myopic_knapsack_oracle": lambda env, obs: oracle_action(env),
}
shock_results = {}
for p_shock in [0.0, 0.20]:
    set_shock_probability(p_shock)
    for name, rule in rules.items():
        if name == "ppo_capacity_aware":
            runs = [mean_over_scenarios(ppo_capacity_aware(ppo_models[s])) for s in TRAIN_SEEDS]
            eps = [e for r in runs for e in r["episodes"]]; mean = float(np.mean([r["mean"] for r in runs]))
        else:
            r = mean_over_scenarios(rule); eps = r["episodes"]; mean = r["mean"]
        rec = {"mean": mean}
        if p_shock > 0:
            sr = [x for e in eps for x, f in zip(e["step_rewards"], e["shocks"]) if f]
            nr = [x for e in eps for x, f in zip(e["step_rewards"], e["shocks"]) if not f]
            rec.update(sprint_reward_with_shock=float(np.mean(sr)), sprint_reward_no_shock=float(np.mean(nr)),
                       n_shock_sprints=len(sr))
        shock_results.setdefault(name, {})[f"p={p_shock}"] = rec
set_shock_probability(0.20)

print(f"{'Method':<26}{'No shocks':>11}{'Shocks 0.2':>12}{'Change':>9}{'Sprint w/ shock':>17}{'Sprint w/o':>12}")
for name, d in shock_results.items():
    a, b = d["p=0.0"]["mean"], d["p=0.2"]["mean"]
    print(f"{name:<26}{a:>11.2f}{b:>12.2f}{b - a:>9.2f}{d['p=0.2']['sprint_reward_with_shock']:>17.2f}{d['p=0.2']['sprint_reward_no_shock']:>12.2f}")
RESULTS_A["shocks"] = shock_results
save_A()
```

### Cell 8. Evidence for the capacity reset bug, then save Part A
Runs the **original** training environment for 98,304 steps with random actions (as SB3 does: one environment, reset between episodes) and records the team capacity at the end of each episode. Random actions are used because the real training trajectories were not logged, so this shows the mechanism, not the exact values each policy saw.

```python
drift = {}
for s in TRAIN_SEEDS:
    env = BacklogPrioritisationEnv(n_items=30, capacity_points=CAPACITY, seed=s); env.reset(seed=s)
    rng = np.random.default_rng(0); caps = []; steps = 0
    while steps < 98_304:
        _, _, _, trunc, _ = env.step(rng.integers(0, 3, size=MAX_ITEMS)); steps += 1
        if trunc: caps.append(env.capacity_points); env.reset()
    caps = np.array(caps)
    drift[s] = dict(median=float(np.median(caps)), max=int(caps.max()), share_ge_50=float(np.mean(caps >= 50)))
    print(f"Seed {s}: capacity at episode end, median {np.median(caps):.0f}, max {caps.max()}, "
          f"episodes with capacity >= 50: {np.mean(caps >= 50):.0%}  (test capacity starts at 20)")
RESULTS_A["capacity_drift_random_actions"] = drift

with open("revision_partA.json", "w") as f: json.dump(RESULTS_A, f, indent=2, default=float)
print("\nSaved revision_partA.json. Part A complete.")
```

## Part B. Fix the capacity reset bug and retrain (about 3 hours, run overnight)

**Before running this cell:** save `REVISION_PROTOCOL.md` in this folder and commit it to Git (see the instructions you were given). That timestamped commit is the evidence that this protocol was fixed before the retraining.

Change only the settings at the top of the cell:
* `RETRAIN_CONFIGS`: by default only the corrected capacity weight (1.0). Add `0.4` to also train the original weight for the reward × decoding comparison (another 3 hours).

```python
class FixedBacklogEnv(BacklogPrioritisationEnv):
    """Identical to the frozen environment except reset() restores the starting team capacity."""
    def __init__(self, n_items=30, capacity_points=20, seed=None):
        super().__init__(n_items=n_items, capacity_points=capacity_points, seed=seed)
        self.initial_capacity = capacity_points
    def reset(self, *, seed=None, options=None):
        self.capacity_points = self.initial_capacity
        return super().reset(seed=seed, options=options)

# Check the fix before spending 3 hours
_e = FixedBacklogEnv(seed=10); _e.reset(seed=10); _r = np.random.default_rng(0); _caps = []
for _ in range(12 * 300):
    _, _, _, _t, _ = _e.step(_r.integers(0, 3, size=MAX_ITEMS))
    if _t: _e.reset(); _caps.append(_e.capacity_points)
assert set(_caps) == {CAPACITY}, "Capacity is not reset. Do not train."
print("Check passed: capacity resets to", CAPACITY, "at the start of every episode.")

from stable_baselines3.common.monitor import Monitor
from stable_baselines3.common.callbacks import BaseCallback

RETRAIN_CONFIGS = [1.0]          # add 0.4 for the factorial comparison, e.g. [1.0, 0.4]
REV_TRAIN_SEEDS = [10, 11, 12, 13, 14, 15, 16, 17]
REV_TIMESTEPS = 98_304

def linear_schedule(initial_value, final_value):
    return lambda progress_remaining: final_value + progress_remaining * (initial_value - final_value)

class ConvergenceLogger(BaseCallback):
    def __init__(self, log_every=2048):
        super().__init__(); self.log_every = log_every; self.history = []
    def _on_step(self):
        if self.n_calls % self.log_every == 0 and len(self.model.ep_info_buffer) > 0:
            self.history.append((self.num_timesteps, float(np.mean([e["r"] for e in self.model.ep_info_buffer]))))
        return True

def model_dir_for(w3): return "models_fixed" if w3 == 1.0 else f"models_fixed_w3_{str(w3).replace('.', '')}"

t_all = time.time()
for w3 in RETRAIN_CONFIGS:
    out_dir = model_dir_for(w3); os.makedirs(out_dir, exist_ok=True)
    for s in REV_TRAIN_SEEDS:
        if os.path.exists(f"{out_dir}/ppo_seed{s}.zip"):
            print(f"[w3={w3}] seed {s} already trained, skipping"); continue
        set_capacity_weight(w3); set_shock_probability(0.20)
        env = Monitor(FixedBacklogEnv(n_items=30, capacity_points=CAPACITY, seed=s))
        model = PPO(policy="MlpPolicy", env=env, learning_rate=linear_schedule(3e-4, 3e-5),
                    n_steps=2048, batch_size=64, gamma=0.99, clip_range=0.2,
                    policy_kwargs=dict(net_arch=dict(pi=[128, 128], vf=[128, 128])),
                    verbose=0, seed=s, device="cpu")
        cb = ConvergenceLogger(); t0 = time.time()
        model.learn(total_timesteps=REV_TIMESTEPS, callback=cb, progress_bar=False)
        model.save(f"{out_dir}/ppo_seed{s}")
        np.save(f"{out_dir}/convergence_seed{s}.npy", np.array(cb.history))
        print(f"[w3={w3}] seed {s} done in {(time.time() - t0) / 60:.1f} min, final training reward {cb.history[-1][1]:.2f}")
set_capacity_weight(1.0)
print(f"\nAll retraining finished in {(time.time() - t_all) / 60:.1f} min.")
```

## Part C. Evaluate the retrained models
All scores use the same test reward as the frozen study (capacity weight 1.0), so every number is comparable. Run Cells 1, 2 and the Part B cell first if you restarted (the Part B cell skips seeds that are already trained).

```python
set_capacity_weight(1.0); set_shock_probability(0.20)
if "det" not in globals() or "controls" not in globals():   # kernel was restarted after Part A
    _A = json.load(open("revision_partA.json"))
    det = _A["sanity"]["det_seed_means"]; cap = _A["sanity"]["cap_seed_means"]; controls = _A["controls"]
RESULTS_C = {"frozen_reference": {"deterministic_mean": float(np.mean(det)), "capacity_aware_mean": float(np.mean(cap)),
                                  "rice_mean": RICE_MEAN, "wsjf_mean": WSJF_MEAN}}
for w3 in RETRAIN_CONFIGS:
    d = model_dir_for(w3)
    models = {s: PPO.load(f"{d}/ppo_seed{s}", device="cpu") for s in REV_TRAIN_SEEDS}
    final_train = np.array([np.load(f"{d}/convergence_seed{s}.npy")[-1][1] for s in REV_TRAIN_SEEDS])
    block = {"final_training_reward": final_train.tolist()}
    print(f"\n==================== Retrained, training capacity weight {w3} ====================")
    for mode, rule in [("deterministic", ppo_deterministic), ("sampled", ppo_sampled), ("capacity_aware", ppo_capacity_aware)]:
        rows = [mean_over_scenarios(rule(models[s]), env_cls=FixedBacklogEnv) for s in REV_TRAIN_SEEDS]
        seed_means = [r["mean"] for r in rows]
        res = compare_to_baselines(f"[w3={w3}] {mode}", seed_means)
        res["n_now_per_sprint"] = float(np.mean([r["n_now"] for r in rows]))
        res["overrun_per_sprint"] = float(np.mean([r["overrun"] for r in rows]))
        r, p = stats.pearsonr(final_train, seed_means); rho, prho = stats.spearmanr(final_train, seed_means)
        res["train_vs_test"] = dict(pearson_r=float(r), pearson_p=float(p), spearman_rho=float(rho), spearman_p=float(prho))
        print(f"  Now items/sprint {res['n_now_per_sprint']:.2f}, overrun {res['overrun_per_sprint']:.2f}; "
              f"training vs test: Pearson r={r:.3f} (p={p:.4f}), Spearman {rho:.3f}")
        block[mode] = res
    block["random_ranking_control"] = controls["random_ranking"]
    RESULTS_C[f"w3={w3}"] = block

with open("revision_partC.json", "w") as f: json.dump(RESULTS_C, f, indent=2, default=float)
print("\nSaved revision_partC.json. Part C complete.")
```
